### 6. Module - Joining and Splitting arrays.


23. Concatenate()

Joins arrays along an existing axis.

```python
import numpy as np
a = np.array([1,2,3])
print(a)
b = np.array([4,5,6])
print(b)
```

```python
print(np.concatenate((a,b)))
```

for two-dimensional array, specify the axis:

```python
a = np.array([
    [1,2],
    [3,4]
])
b = np.array([
    [5,6],
    [7,8]
])
print(a)
print(b)
```

```python
print(np.concatenate((a,b),axis=0))
```

```python
a = np.array([
    [1,2],
    [3,4]
])
b = np.array([
    [5,6],
    [7,8]
])
print(np.concatenate((a,b),axis=1))
```

### 24. `stack(),hstack() and vstack()`

stack() joins arrays along a new axis

```python
a = np.array([1,2])
b = np.array([3,4])
print(a)
print(b)
```

```python
print(np.stack((a,b),axis=0)) # rows and columns
```

Horizonal and vertical stacking

```python
print(np.hstack((a,b))) # horizonal - columns
```

```python
print(np.vstack((a,b)))  # Vertical stack- rows
```
