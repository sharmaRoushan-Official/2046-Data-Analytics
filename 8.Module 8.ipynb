28. where()

Selects values according to a condition.

```python
import numpy as np
marks = np.array([35,78,45,90,28])

result = np.where(marks>=40,"Pass", "Fail")
print(result)
```

### It can also return the indices where a condition is true:

```python
indexValue = np.where(marks>=40)
print(indexValue)
```

### 29. sort() and argsort()

np.sort return sorted values

```python
arr= np.array([50,20,40,10,30])  # unsorted
print("Original Array:",arr)
print("Sorted Array",np.sort(arr))
```

##### np.argsort() returns the indices that would sort the array

```python
print(np.argsort(arr))
```

These indices corresponds to values `10,20,30,40,50` in the original array


### 30. `argmax()` and `argmin()`



#### Return the index of maximin and mininum

```python
marks = np.array([65,90,45,78])
print(max(marks))  # value
print(np.argmax(marks))  # index value

print(min(marks))  # value
print(np.argmin(marks)) # index value
```

Notes:

Use `max()` and `min` when you need values, and
`argmax()` and `argmin()` when you need positions.
