# NumPy: Generate a 1D Array of 5 Random Floats in [10.0, 20.0)

## 1. Python Code

```python
import numpy as np

rng = np.random.default_rng(seed=42)

# Generate a 1D array of 5 floats in [10.0, 20.0)
array_1d = rng.uniform(low=10.0, high=20.0, size=5)

print(array_1d)
```

## 2. Output

```text
[17.73956049 14.3887844  18.5859792
 16.97368029 10.94177348]
```

## 3. Explanation of Parameters

| Parameter | Value | Meaning |
|---|---|---|
| `rng` | Generator | Random number generator |
| `low` | 10.0 | Minimum boundary |
| `high` | 20.0 | Maximum boundary (excluded) |
| `size` | 5 | Generates 5 random numbers |
| `uniform()` | — | Generates uniformly distributed floating-point numbers |

## 4. Mathematical Formula

NumPy uses the following formula:

\[
X=low+(high-low)\times U
\]

Where:

- \(X\) = Generated random number
- \(low\) = Lower boundary
- \(high\) = Upper boundary
- \(U\) = Random number between 0 and 1

### Example: Calculate the First Random Number

Suppose:

\[
U=0.773956049
\]

Given:

- Low = 10
- High = 20

Calculation:

\[
X=10+(20-10)\times0.773956049
\]

\[
X=10+10\times0.773956049
\]

\[
X=10+7.73956049
\]

\[
\boxed{X=17.73956049}
\]

## 5. Understand All Five Values

| Random U | Calculation | Final Value |
|---|---|---|
| 0.773956049 | 10 + 10 × 0.773956049 | 17.73956049 |
| 0.438878440 | 10 + 10 × 0.438878440 | 14.38878440 |
| 0.858597920 | 10 + 10 × 0.858597920 | 18.58597920 |
| 0.697368029 | 10 + 10 × 0.697368029 | 16.97368029 |
| 0.094177348 | 10 + 10 × 0.094177348 | 10.94177348 |

## 6. Important Concept: Why Does the Output Change?

The random generator maintains its internal state.

- Every time we generate random numbers, the generator advances its sequence.
- Using the same seed with a fresh generator produces the same sequence.
- Using an existing generator produces the next values in its sequence.

Example:

```python
rng = np.random.default_rng(seed=42)

# First generation
a = rng.uniform(0, 10, size=5)

# Second generation
b = rng.uniform(10, 20, size=5)
```

The second generation continues from the generator's updated state.

## 7. Quick Revision for Exams

| Code | Meaning |
|---|---|
| `rng.uniform(0, 1, 5)` | 5 random floats between 0 and 1 |
| `rng.uniform(10, 20, 5)` | 5 random floats between 10 and 20 |
| `rng.uniform(1, 100, 10)` | 10 random floats between 1 and 100 |

**Remember:**

- `low` → Starting boundary.
- `high` → Ending boundary (excluded).
- `size` → Number of random values.
- `uniform()` → Generates random floating-point values.
- `default_rng(seed=42)` → Creates a reproducible random generator.

````python
import numpy as np

# 1. Initialize the random number generator (pass an optional seed for reproducibility)
rng = np.random.default_rng(seed=42)
print(rng)

# 2. Generate a single random float in [0.0, 1.0)
single_val = rng.uniform()
print(single_val)

# 3. Generate a 1D array of 5 floats in [10.0, 20.0)
array_1d = rng.uniform(low=10.0, high=20.0, size=5)
print(array_1d)

# 4. Generate a 2x3 matrix of floats in [-1.0, 1.0)
matrix_2d = rng.uniform(low=-1.0, high=1.0, size=(2, 3))
print(matrix_2d)

matrix_3d = rng.uniform(low=-1.0, high=1.0, size=(2, 2, 3))
print(matrix_3d)

````
