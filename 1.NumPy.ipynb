## Signed and Unsigned Integer Data Types

| Category | NumPy Type | String Code | Bits | Range |
|---|---|---|---:|---:|
| **Signed** | `np.int8` | `i1` | 8-bit | −128 to 127 |
| **Signed** | `np.int16` | `i2` | 16-bit | −32,768 to 32,767 |
| **Signed** | `np.int32` | `i4` | 32-bit | −2,147,483,648 to 2,147,483,647 |
| **Signed** | `np.int64` | `i8` | 64-bit | −9,223,372,036,854,775,808 to 9,223,372,036,854,775,807 |
| **Unsigned** | `np.uint8` | `u1` | 8-bit | 0 to 255 |
| **Unsigned** | `np.uint16` | `u2` | 16-bit | 0 to 65,535 |
| **Unsigned** | `np.uint32` | `u4` | 32-bit | 0 to 4,294,967,295 |
| **Unsigned** | `np.uint64` | `u8` | 64-bit | 0 to 18,446,744,073,709,551,615 |

### Quick Rule

| Type | Meaning | Formula for Range |
|---|---|---|
| **Signed n-bit** | Positive + Negative | `−2^(n−1)` to `2^(n−1) − 1` |
| **Unsigned n-bit** | Positive only | `0` to `2^n − 1` |

**Example:**  
For `int8` → `−2⁷` to `2⁷ − 1` → **−128 to 127**  
For `uint8` → `0` to `2⁸ − 1` → **0 to 255**

````python
import numpy as np

# Signed integers
a = np.array([100], dtype=np.int8)
b = np.array([30000], dtype=np.int16)
c = np.array([2000000000], dtype=np.int32)
d = np.array([9000000000000000000], dtype=np.int64)

# Unsigned integers
e = np.array([255], dtype=np.uint8)
f = np.array([60000], dtype=np.uint16)
g = np.array([4000000000], dtype=np.uint32)
h = np.array([18000000000000000000], dtype=np.uint64)

print(a)
print(a.dtype)
print(b)
print(b.dtype)
print(c)
print(c.dtype)
print(d)
print(d.dtype)
print(e)
print(e.dtype)
print(f)
print(f.dtype)
print(g)
print(g.dtype)
print(h)
print(h.dtype)
````

## Population Variance

Here is the formula and required steps for finding the variance of a population:

### Variance for a Population

| Formula | Calculation Steps |
|:---|:---|
| $$\sigma^2 = \frac{\sum (x-\mu)^2}{N}$$ | 1. Find the mean. <br> 2. Subtract the mean from each raw score to find each deviation. <br> 3. Square each deviation. <br> 4. Sum the squared deviations to find SS. <br> 5. Divide SS by $N$. |


## Example: Population Variance

Suppose the marks of 3 students are **2, 4, 6**. Find the population variance.

### Step 1: Find the Mean

$$
\mu = \frac{2+4+6}{3} = \frac{12}{3} = 4
$$

### Step 2: Subtract the Mean from Each Value

| Value ($x$) | Deviation ($x-\mu$) |
|---:|---:|
| 2 | $2-4=-2$ |
| 4 | $4-4=0$ |
| 6 | $6-4=2$ |

### Step 3: Square Each Deviation

| Deviation ($x-\mu$) | Squared Deviation $(x-\mu)^2$ |
|---:|---:|
| -2 | 4 |
| 0 | 0 |
| 2 | 4 |
| **Total** | **8** |

### Step 4: Apply the Population Variance Formula

$$
\sigma^2 = \frac{\sum(x-\mu)^2}{N}
$$

$$
\sigma^2 = \frac{8}{3} = 2.67
$$

### Final Answer

**Population Variance = 2.67**

**Remember:** Mean → Subtract → Square → Add → Divide by $N$.


## Example: Calculate Standard Deviation Step by Step

Given data:

$$
x = [2, 4, 6]
$$

### Step 1: Find the Mean

$$
\mu = \frac{\sum x}{N}
$$

$$
\mu = \frac{2+4+6}{3} = \frac{12}{3} = 4
$$

### Step 2: Calculate Deviations

Subtract the mean from each value.

| Value ($x$) | Calculation ($x-\mu$) | Deviation |
|---:|---:|---:|
| 2 | $2-4$ | -2 |
| 4 | $4-4$ | 0 |
| 6 | $6-4$ | 2 |

### Step 3: Square Each Deviation

| Deviation ($x-\mu$) | Squared Deviation $(x-\mu)^2$ |
|---:|---:|
| -2 | 4 |
| 0 | 0 |
| 2 | 4 |
| **Total** | **8** |

$$
\sum(x-\mu)^2 = 8
$$

### Step 4: Calculate Population Variance

Formula:

$$
\sigma^2 = \frac{\sum(x-\mu)^2}{N}
$$

Substitute the values:

$$
\sigma^2 = \frac{8}{3} = 2.6667
$$

### Step 5: Calculate Standard Deviation

Standard deviation is the square root of variance.

$$
\sigma = \sqrt{\frac{\sum(x-\mu)^2}{N}}
$$

$$
\sigma = \sqrt{\frac{8}{3}}
$$

$$
\sigma = \sqrt{2.6667} \approx 1.63299
$$

### Final Answer

**Population Standard Deviation = 1.63299**

# Quartiles

A **quartile** divides sorted data into four equal parts. Each part represents approximately 25% of the data.

## 1. Types of Quartiles

- **Q1 (First Quartile):** 25th percentile
- **Q2 (Second Quartile):** 50th percentile (Median)
- **Q3 (Third Quartile):** 75th percentile

## 2. Quartile Position Formulas

One commonly taught method uses the following formulas:

### First Quartile (Q1)

$$
\text{Position of } Q_1 = \frac{N+1}{4}
$$

### Second Quartile (Q2)

$$
\text{Position of } Q_2 = \frac{2(N+1)}{4}
$$

### Third Quartile (Q3)

$$
\text{Position of } Q_3 = \frac{3(N+1)}{4}
$$

**Where:**

- $N$ = Total number of data values
- $Q_1$ = First quartile
- $Q_2$ = Second quartile
- $Q_3$ = Third quartile

**Note:** These formulas calculate quartile positions. If a position is fractional, interpolation may be needed.

## 3. Example: Calculate Q1, Q2 and Q3

Given data:

$$
10, 20, 30, 40, 50, 60, 70
$$

### Step 1: Count the Values

$$
N = 7
$$

### Step 2: Calculate Q1 Position

$$
Q_1 = \frac{N+1}{4}
$$

$$
Q_1 = \frac{7+1}{4} = \frac{8}{4} = 2
$$

The 2nd value is 20.

$$
\boxed{Q_1 = 20}
$$

### Step 3: Calculate Q2 Position

$$
Q_2 = \frac{2(N+1)}{4}
$$

$$
Q_2 = \frac{2(7+1)}{4} = \frac{16}{4} = 4
$$

The 4th value is 40.

$$
\boxed{Q_2 = 40}
$$

### Step 4: Calculate Q3 Position

$$
Q_3 = \frac{3(N+1)}{4}
$$

$$
Q_3 = \frac{3(7+1)}{4} = \frac{24}{4} = 6
$$

The 6th value is 60.

$$
\boxed{Q_3 = 60}
$$

# Normal Distribution

## 1. Introduction

Normal Distribution is a continuous probability distribution that is symmetric and bell-shaped.

In a normal distribution:

- Mean = Median = Mode
- The curve is symmetric around the mean.
- Most observations are concentrated around the mean.
- The tails extend towards both sides.

## 2. Normal Distribution Formula (PDF)

The Probability Density Function (PDF) is:

$$
f(x) = \frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

### Where:

- $x$ = Individual data value
- $\mu$ = Mean
- $\sigma$ = Standard deviation
- $\sigma^2$ = Variance
- $e$ = Euler's number (2.71828)
- $\pi$ = 3.14159

## 3. Standard Normal Distribution (Z-Score)

The Z-score measures how many standard deviations an observation is away from the mean.

### Formula:

$$
Z = \frac{X-\mu}{\sigma}
$$

Where:

- $X$ = Individual observation
- $\mu$ = Mean
- $\sigma$ = Standard deviation

For standard normal distribution:

$$
\mu = 0,\quad \sigma = 1
$$

## 4. Empirical Rule (68-95-99.7 Rule)

The empirical rule describes how data is distributed around the mean.

| Range | Percentage |
|---|---:|
| $\mu \pm 1\sigma$ | 68% |
| $\mu \pm 2\sigma$ | 95% |
| $\mu \pm 3\sigma$ | 99.7% |

## 5. Solved Numerical Example

**Question:**

Suppose students' marks follow a normal distribution:

- Mean = 60
- Standard Deviation = 10
- Student's Marks = 80

Calculate the Z-score.

### Step 1: Apply the formula

$$
Z = \frac{X-\mu}{\sigma}
$$

### Step 2: Substitute the values

$$
Z = \frac{80-60}{10}
$$

### Step 3: Calculate

$$
Z = \frac{20}{10}
$$

$$
\boxed{Z=2}
$$

**Interpretation:** The student's marks are 2 standard deviations above the mean.

A Uniform Distribution is a probability distribution in which all values or outcomes within a specified range have equal probability density.

- $x$ = Value between $a$ and $b$

## 3. Binomial Distribution

A Binomial Distribution is a discrete probability distribution that represents the number of successes in a fixed number of independent trials.

Each trial has only two possible outcomes:

- Success
- Failure

### Characteristics

- Fixed number of trials (n).
- Only two possible outcomes per trial.
- Trials are independent.
- Probability of success (p) remains constant.

### Formula

$$
P(X=k) = {n \choose k}p^k(1-p)^{n-k}
$$

Where:

- $n$ = Total number of trials
- $k$ = Number of successes
- $p$ = Probability of success
- $1-p$ = Probability of failure

### Example

A coin is tossed 5 times.

Find the probability of getting exactly 3 heads.

Given:

- $n = 5$
- $k = 3$
- $p = 0.5$

$$
P(X=3) = {5 \choose 3}(0.5)^3(0.5)^2
$$

$$
= 10 \times 0.125 \times 0.25
$$

$$
\boxed{P(X=3)=0.3125}
$$

Probability = 31.25%.

### NumPy Example

## 4. Poisson Distribution

A Poisson Distribution is a discrete probability distribution used to calculate the probability of a specific number of events occurring within a fixed interval of time or space.

### Characteristics

- Counts the number of events.
- Events occur independently.
- Average event rate remains constant.
- Events occur individually over a specified interval.

### Formula

$$
P(X=k) = \frac{e^{-\lambda}\lambda^k}{k!}
$$

Where:

- $k$ = Number of events
- $\lambda$ = Average number of events
- $e$ = Euler's number (2.71828)
- $k!$ = Factorial of k

### Example

A call center receives an average of 4 calls per minute.

Find the probability of receiving exactly 2 calls in one minute.

Given:

- $\lambda = 4$
- $k = 2$

$$
P(X=2) = \frac{e^{-4}\times4^2}{2!}
$$

$$
= \frac{0.018315 \times 16}{2}
$$

$$
\boxed{P(X=2)\approx0.1465}
$$

Probability ≈ 14.65%.

### NumPy Example

```python
import numpy as np

plt.show()
```

## 5. Exponential Distribution

An Exponential Distribution is a continuous probability distribution used to model the waiting time between consecutive events occurring at a constant average rate.

### Characteristics

- Continuous distribution.
- Used to calculate waiting time.
- Has a decreasing probability density curve.
- It is memoryless.
- Often used in reliability and queuing systems.

### Formula

Probability Density Function (PDF):

$$
f(x) = \lambda e^{-\lambda x}
$$

For $x \geq 0$.

Cumulative Distribution Function (CDF):

$$
F(x) = 1-e^{-\lambda x}
$$

Where:

- $\lambda$ = Event rate
- $x$ = Waiting time
- $e$ = Euler's number

### Example

A customer arrives at a service counter at an average rate of 2 customers per minute.

Find the probability that the waiting time for the next customer is more than 1 minute.

Given:

- $\lambda = 2$
- $x = 1$

$$
P(X>1)=e^{-\lambda x}
$$

$$
=e^{-2\times1}
$$

$$
\boxed{P(X>1)\approx0.1353}
$$

Probability ≈ 13.53%.

### NumPy Example

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.exponential(scale=1/2, size=1000)

plt.hist(data, bins=30, density=True)
plt.title("Exponential Distribution")
plt.show()
```

Note: In NumPy, `scale` represents the mean waiting time (1/λ), not the rate λ.

# Normal Distribution – 10 Practice Questions

**Topic:** Normal Distribution (Z-score, Z-table and Probability)

**Instructions:** Solve all questions using the formula:

$$
Z = \frac{X-\mu}{\sigma}
$$

Use the Standard Normal Z-table to calculate probabilities.

---

## Section A: Basic Questions

### Q1. Students' Marks

The marks of students follow a normal distribution with mean 50 and standard deviation 10.

Find the Z-score for a student who scored 70 marks.

### Q2. Employee Salary

The average salary of employees is ₹30,000 with a standard deviation of ₹5,000.

Find the Z-score for an employee earning ₹40,000.

### Q3. Students' Height

The average height of students is 170 cm with a standard deviation of 5 cm.

Find the Z-score for a student whose height is 180 cm.

---

## Section B: Probability Calculation

### Q4. Examination Marks

The marks of 1,000 students follow a normal distribution.

Given:

- Mean = 60
- Standard Deviation = 10
- Total Students = 1,000

Calculate the percentage of students scoring between 50 and 70.

### Q5. Student Performance

The average marks are 70 and the standard deviation is 10.

Calculate the percentage of students scoring between 60 and 80.

### Q6. Manufacturing Company

The average weight of a product is 500 grams with a standard deviation of 20 grams.

Find the probability that a randomly selected product weighs between 480 and 520 grams.

### Q7. Employee Age

The average age of employees is 35 years with a standard deviation of 5 years.

Calculate the probability that an employee is older than 40 years.

---

# Formula Cheat Sheet

| Calculation | Formula |
|---|---|
| Z-score | $Z=\frac{X-\mu}{\sigma}$ |
| Below a value | $P(Z<z)=\Phi(z)$ |
| Above a value | $P(Z>z)=1-\Phi(z)$ |
| Between two values | $P(a<Z<b)=\Phi(b)-\Phi(a)$ |
| Number of students | $\text{Probability}\times N$ |
