## 10.ndim()

ndim tells us the number of dimesion


````python
a = np.array([1,2,3])

b = np.array([
    [1,2],
    [3,4]
])
print(a.ndim)  # 1-D
print(b.ndim) # 2-D
````

### 11. shape

shape tells use the size of each dimension



````python
arr = np.array([
    [10,20,30,40],
    [40,50,60,70],
    [70,80,90,100]
])
print(arr.ndim) # 2
print(arr.shape) # 
````

```

```

## 12. size

size returns total number of elements

````python
arr = np.array([
    [10,20,30,40],
    [40,50,60,70],
    [70,80,90,100]
])
print(arr.ndim) # 2
print(arr.shape) # 3,4 
print(arr.size) # 12
````

### 13. dtype

dtype tells us data type of array elements.

    

````python
arr = np.array([10],dtype=np.int64)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
````

## Signed and Unsigned Integer Data Types

| Category | NumPy Type | String Code | Bits | Range |
|---|---|---|---:|---:|
| **Signed** | `np.int8` | `i1` | 8-bit | −128 to 127 |
| **Signed** | `np.int16` | `i2` | 16-bit | −32,768 to 32,767 |
| **Signed** | `np.int32` | `i4` | 32-bit | −2,147,483,648 to 2,147,483,647 |
| **Signed** | `np.int64` | `i8` | 64-bit | −9,223,372,036,854,775,808 to 9,223,372,036,854,775,807 |
| **Unsigned** | `np.uint8` | `u1` | 8-bit | 0 to 255 |
| **Unsigned** | `np.uint16` | `u2` | 16-bit | 0 to 65,535 |
| **Unsigned** | `np.uint32` | `u4` | 32-bit | 0 to 4,294,967,295 |
| **Unsigned** | `np.uint64` | `u8` | 64-bit | 0 to 18,446,744,073,709,551,615 |

### Quick Rule

| Type | Meaning | Formula for Range |
|---|---|---|
| **Signed n-bit** | Positive + Negative | `−2^(n−1)` to `2^(n−1) − 1` |
| **Unsigned n-bit** | Positive only | `0` to `2^n − 1` |

**Example:**  
For `int8` → `−2⁷` to `2⁷ − 1` → **−128 to 127**  
For `uint8` → `0` to `2⁸ − 1` → **0 to 255**

````python
import numpy as np

# Signed integers
a = np.array([100], dtype=np.int8)
b = np.array([30000], dtype=np.int16)
c = np.array([2000000000], dtype=np.int32)
d = np.array([9000000000000000000], dtype=np.int64)

# Unsigned integers
e = np.array([255], dtype=np.uint8)
f = np.array([60000], dtype=np.uint16)
g = np.array([4000000000], dtype=np.uint32)
h = np.array([18000000000000000000], dtype=np.uint64)

print(a)
print(a.dtype)
print(b)
print(b.dtype)
print(c)
print(c.dtype)
print(d)
print(d.dtype)
print(e)
print(e.dtype)
print(f)
print(f.dtype)
print(g)
print(g.dtype)
print(h)
print(h.dtype)
````
