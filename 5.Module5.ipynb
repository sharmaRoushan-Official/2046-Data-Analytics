#### 20. `flatten() vs ravel()`

Both convert a multidimensional array into one dimension, but their memeory behavior differ.



```python
arr = np.array([
    [1,2],
    [3,4]
])
a = arr.flatten()
print(a)
b = arr.ravel()
print(b)
```

Notes:

* flatten() returns a copy.
* ravel() returns a view when possible otherwise a copy.

```python
arr = np.array([
    [1,2],
    [3,4]
])


a = a.flatten()


a[0] = 100
print(arr) # original remains unchaged - [create a copy]
 

b = arr.ravel() 
b[0]=200
print(arr) ## Original may change becuuse b is a view - [original modified]
```

### 21. Transpose
Trnaspose swaps the axes of an array

```python
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr)  # 2 rows and 3 columns
print(arr.shape)
```

```python
arr1 = np.array([
    [1,2,3],
    [4,5,6]
])
x = arr1.T
print(x)
print(x.shape)
```

Note:
For a two-dimensional array, `.T` is a convenient way to exchange rows and columns.

### 22. `squeeze()` and `expand_dims()`


`squeeze()` removes axes of length 1

```python
arr = np.array([[[10,20,30]]])
print(arr)
print(arr.shape) 
print(arr.ndim) # dimension 3 
print(np.squeeze(arr)) 
print(np.squeeze(arr).shape) 
print(np.squeeze(arr).ndim) # dimension 1
```

expand_dim() adds a dimension at a specified position.

```python
arr = np.array([10,20,30])
print(arr.shape) # 1-D array
print(np.expand_dims(arr,axis=0))  # converted into rows and columns - 2-D
print(np.expand_dims(arr,axis=0).shape) # 

print(np.expand_dims(arr,axis=1))
print(np.expand_dims(arr,axis=1).shape) # 3 rows and 1 columns
```

Notes:

These operations are partcularly useful when matching shapes for machine-learning models and broadcasting.
